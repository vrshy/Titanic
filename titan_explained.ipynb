# FILLING NULL VALUES

```python
df
```

```python
df['age'].mean()
```

### 💡 What just happened?
We calculated the **average (mean) age** of all passengers who have an age value.  
We'll use this to fill in the missing ages — a common trick called **mean imputation**.

```python
age_mean=df['age'].mean()
```

### 💡 What just happened?
We saved that average age into a variable called `age_mean` so we can reuse it easily.

```python
df
```

# *types of data*  
1. numerical data → which is a number  
2. categorical data → which is a label/category (like city name, gender, etc.)

```python
df['fare'].mean()
```

### 💡 What just happened?
Same idea — we found the **average ticket fare** to fill in any missing fare values.

```python
fare_mean=df['fare'].mean()
```

### 💡 Saved the average fare into `fare_mean` for later use.

```python
df['embarked'].unique()
```

### 💡 What just happened?
`unique()` shows all the **different values** in the embarked column (e.g., 'S', 'C', 'Q').  
We need to see what values exist before we decide how to fill missing ones.

```python
embarked_mode=df['embarked'].mode()[0]
```

### 💡 What just happened?
`mode()` finds the **most frequently appearing value** — the most common boarding point.  
`[0]` picks the first/top result (in case there's a tie).  
We use mode for text/category columns (not mean, because you can't average letters!).

```python
df['age']=df['age'].fillna(age_mean)
df['fare']=df['fare'].fillna(fare_mean)
df['embarked']=df['embarked'].fillna(embarked_mode)
```

### 💡 What just happened?
`fillna()` fills **all empty (NaN) spots** in a column with the value we give it.
- Empty age → filled with average age
- Empty fare → filled with average fare
- Empty embarked → filled with most common boarding point

Now no more blanks in these 3 columns! ✅

```python
df
```

```python
df.info()
```

### 💡 What just happened?
We check `info()` again to confirm that **non-null counts went up** for age, fare, and embarked.  
If they now show the full count (e.g., 891), all nulls are filled!

## CUSTOM FUNCTION


```python
def comp_type(row):
    f=row['fam_cnt']
    p=row['no of psgs']
    if p==1:
        return 'solo'
    elif f==0 and p>1:
        return 'friends'
    else:
        return 'family' 
```

### 💡 What just happened?
We wrote a **custom function** `comp_type` that looks at each passenger and decides:
- If only 1 person on the ticket → `'solo'`  
- If multiple people on ticket but no family members → `'friends'`  
- Otherwise → `'family'`

This function will be applied to every row one by one.

```python
df
```

```python
df.apply(comp_type,axis=1)
```

### 💡  APPLY()
- `df.apply(comp_type, axis=1)` **runs our function on every single row**.  
- `axis=1` means go row by row (not column by column).                                                             
- `comp_type` Is user defined.                                 
- The result is a new Series showing solo/friends/family for each passenger.

```python
df['companion_type']=df.apply(comp_type,axis=1)
```

### 💡 What just happened?
We saved the result into a new column `companion_type`.  
Now every passenger is labelled as `'solo'`, `'friends'`, or `'family'` — very useful for analysis!

```python
df
```

### 💡 CUT()
`pd.cut()` puts each age into a **bucket (bin)** and gives it a label.  
The bins define the age ranges:
- 0–2 → `'infant'`
- 2–16 → `'child'`
- 16–25 → `'teen'`
- 25–59 → `'adult'`
- 59–100 → `'senior'`                               
            it can be used only numerical data we cant use it for categorical data.                   
            `bins[0,2,16,25,59,100]`          
            here bins`[0,2]`0 is excluded.2 is included.     
            bins`[2,16]` 2 is excluded 16 is included.    
- like this it will compare 2 2 numbers by including and excluding

This is called **binning** — turning a continuous number into a category. Super useful in ML!

```python

df['age category']=pd.cut(df['age'],bins=[0,2,16,25,59,100],labels=['infant','child','teen','adult','senior'])
#def age_cat(res):
    #if age>=0 and age<=2:
        #return 'infant'
    #elif age>2 and age <=16:
        #return 'child'
    #elif age>16 and age<=25:
        #return 'teen'
    #elif age>25 and age<=59:
        #return 'adult'
    #elif age>59 and age<=100:
        #return "senior"

 
```
